```python
# Q.1
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)

print("a) After adding 2:", arr + 2)
print("b) After multiplying by 3:", arr * 3)
print("c) After dividing by 2:", arr / 2)
```

```python
# Q.2
import numpy as np

# a)
arr = np.array([1, 2, 3, 6, 4, 5])
rev_arr = arr[::-1]
print("a) Reversed array:", rev_arr)

# b) i.
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print("b) i. Most frequent value in x:", most_freq_x, "at indices:", indices_x)

# b) ii.
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_freq_y = vals_y[counts_y == max_count_y]
print("b) ii. Most frequent value(s) in y:")
for val in most_freq_y:
    print("Value:", val, "at indices:", np.where(y == val)[0])
```

```python
# Q.4
import numpy as np

Aayaan_Dua = np.linspace(10, 100, 25)
print("Array:", Aayaan_Dua)
print("Dimensions:", Aayaan_Dua.ndim)
print("Shape:", Aayaan_Dua.shape)
print("Total elements:", Aayaan_Dua.size)
print("Data type:", Aayaan_Dua.dtype)
print("Total bytes consumed:", Aayaan_Dua.nbytes)

transposed_arr = Aayaan_Dua.reshape(25, 1)
print("Transpose using reshape():\n", transposed_arr)

print("Using .T on 1D array:", Aayaan_Dua.T)
print("No, .T attribute does not transpose a 1D array in NumPy; it returns the exact same 1D shape (25,).")
```

```python
# Q.5
import numpy as np

ucs420_aayaan_dua = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("Original 3x4 array:\n", ucs420_aayaan_dua)

print("Mean:", np.mean(ucs420_aayaan_dua))
print("Median:", np.median(ucs420_aayaan_dua))
print("Max:", np.max(ucs420_aayaan_dua))
print("Min:", np.min(ucs420_aayaan_dua))
print("Unique elements:", np.unique(ucs420_aayaan_dua))

reshaped_ucs420_aayaan_dua = ucs420_aayaan_dua.reshape(4, 3)
print("Reshaped 4x3 array:\n", reshaped_ucs420_aayaan_dua)

resized_ucs420_aayaan_dua = np.resize(ucs420_aayaan_dua, (2, 3))
print("Resized 2x3 array:\n", resized_ucs420_aayaan_dua)
```
