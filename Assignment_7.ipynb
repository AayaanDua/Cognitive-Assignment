```python
# Q.1 - Part I
import numpy as np
import pandas as pd

np.random.seed(1024170284)

sales_data = np.random.randint(1000, 5001, size=(12, 4))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
categories = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]

df = pd.DataFrame(sales_data, columns=categories, index=months)
print(df)
```

```python
# Q.1 - Part II
print("First 5 rows:")
print(df.head())

print("\nSummary statistics:")
print(df.describe())

total_per_category = df[categories].sum()
print("\nTotal sales per category:")
print(total_per_category)

total_per_month = df[categories].sum(axis=1)
print("\nTotal sales per month:")
print(total_per_month)

avg_growth_category = df[categories].pct_change().mean() * 100
print("\nAverage monthly sales growth (%) per category:")
print(avg_growth_category)

df["Total Sales"] = df[categories].sum(axis=1)
df["Growth Rate"] = df["Total Sales"].pct_change() * 100

roll_no = 1024170284
if roll_no % 2 == 0:
    df["Electronics"] = df["Electronics"] * 0.90
else:
    df["Clothing"] = df["Clothing"] * 0.85

print("\nUpdated DataFrame:")
print(df)
```

```python
# Q.1 - Part III
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
for col in categories:
    plt.plot(df.index, df[col], marker="o", label=col)

plt.title("Monthly Sales Trends by Category")
plt.xlabel("Month")
plt.ylabel("Sales Units")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df[categories], palette="Set2")
plt.title("Sales Distribution by Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Units")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
# Q.2
import numpy as np

array = np.array([[1, 2, 3], [-4, 5, -6]])

# i.
abs_array = np.abs(array)
print("i. Element-wise absolute value:\n", abs_array)

# ii.
print("\nii. Percentiles (25th, 50th, 75th):")
print("Flattened array:", np.percentile(array, [25, 50, 75]))
print("Each column:\n", np.percentile(array, [25, 50, 75], axis=0))
print("Each row:\n", np.percentile(array, [25, 50, 75], axis=1))

# iii.
print("\niii. Mean, Median and Standard Deviation:")
print("Flattened -> Mean:", np.mean(array), "| Median:", np.median(array), "| Std:", np.std(array))
print("Column-wise -> Mean:", np.mean(array, axis=0), "| Median:", np.median(array, axis=0), "| Std:", np.std(array, axis=0))
print("Row-wise -> Mean:", np.mean(array, axis=1), "| Median:", np.median(array, axis=1), "| Std:", np.std(array, axis=1))
```

```python
# Q.3
import numpy as np

a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

# i.
print("Original array:", a)
print("Floor values:", np.floor(a))
print("Ceiling values:", np.ceil(a))
print("Truncated values:", np.trunc(a))
print("Rounded values:", np.round(a))
```
