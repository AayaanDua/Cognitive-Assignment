## Q2. Basic NumPy array questions

### Q2 (a). Reverse the given NumPy array

```python
arr = np.array([1, 2, 3, 6, 4, 5])

reversed_arr = arr[::-1]

print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

### Q2 (b). Find the most frequent value and its index/indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values_x, counts_x = np.unique(x, return_counts=True)
most_frequent_x = values_x[counts_x == counts_x.max()]
indices_x = np.where(np.isin(x, most_frequent_x))[0]

values_y, counts_y = np.unique(y, return_counts=True)
most_frequent_y = values_y[counts_y == counts_y.max()]
indices_y = np.where(np.isin(y, most_frequent_y))[0]

print("For array x:")
print("Most frequent value(s):", most_frequent_x)
print("Index/indices:", indices_x)

print("\nFor array y:")
print("Most frequent value(s):", most_frequent_y)
print("Index/indices:", indices_y)
```

## Q4. Create an array with 25 evenly spaced numbers from 10 to 100

```python
Aayaan_Dua = np.linspace(10, 100, 25)

print("Array:", Aayaan_Dua)
print("Dimensions:", Aayaan_Dua.ndim)
print("Shape:", Aayaan_Dua.shape)
print("Total elements:", Aayaan_Dua.size)
print("Data type:", Aayaan_Dua.dtype)
print("Total bytes:", Aayaan_Dua.nbytes)

reshaped_Aayaan_Dua = Aayaan_Dua.reshape(5, 5)
print("\nArray after reshape:")
print(reshaped_Aayaan_Dua)

print("\nTranspose using T:")
print(Aayaan_Dua.T)

print("\nTranspose after reshaping into a 2-D array:")
print(reshaped_Aayaan_Dua.T)
```

**Answer:** A one-dimensional array does not change when we use the `T` attribute because its transpose is the same one-dimensional array. To get a visible transpose, we first reshape it into a two-dimensional array and then use `.T`.

## Q5. Create a 2-D array and calculate statistical values, reshape and resize it

```python
ucs420_Aayaan_Dua = np.array([[10, 20, 30, 40],
                              [50, 60, 70, 80],
                              [90, 15, 20, 35]])

print("Original array:")
print(ucs420_Aayaan_Dua)

print("\nMean:", np.mean(ucs420_Aayaan_Dua))
print("Median:", np.median(ucs420_Aayaan_Dua))
print("Maximum:", np.max(ucs420_Aayaan_Dua))
print("Minimum:", np.min(ucs420_Aayaan_Dua))
print("Unique elements:", np.unique(ucs420_Aayaan_Dua))

reshaped_ucs420_Aayaan_Dua = ucs420_Aayaan_Dua.reshape(4, 3)
print("\nReshaped array (4 rows and 3 columns):")
print(reshaped_ucs420_Aayaan_Dua)

resized_ucs420_Aayaan_Dua = np.resize(ucs420_Aayaan_Dua, (2, 3))
print("\nResized array (2 rows and 3 columns):")
print(resized_ucs420_Aayaan_Dua)
```
